```python
import pandas as pd

df = pd.read_csv("customers.csv")
df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df["data_gb"] = df["data_gb"].fillna(df["data_gb"].median())
df["voice_min"] = df["voice_min"].fillna(df["voice_min"].median())
```

```python
df.isna().sum()
```

```python
df["data_gb"].describe()
```

```python
cap = df["data_gb"].quantile(0.99)
# quantile(0.99) finds the value that 99% of customers are below.
print("Cap value:", cap)
```

```python
df["data_gb"] = df["data_gb"].clip(upper=cap)
df["data_gb"].max()
#This line goes through all 5,000 numbers and applies one rule: "nobody may be above 59.7."
# A customer with 30 GB stays at 30 GB.
# A customer with 252.5 GB becomes 59.7 GB.
```

```python
df.to_csv("customers_clean.csv", index=False)
```

```python
pd.read_csv("customers_clean.csv").shape
```

```python
# Step 1: choose which columns the model will use
features = ["data_gb", "voice_min", "sms_count",
            "monthly_spend_da", "topup_count", "tenure_months"]

X = df[features]
X.head()
```

```python
#K-Means measures distance between customers. Spend has much bigger numbers than data, so a difference of 500 DA would look huge, 
#and a difference of 10 GB would look tiny. 
#The model would listen almost only to spend and ignore everything else.
#It's like comparing a person's height in millimeters with their age in years. The millimeters would always win.

#Scaling fixes this. It converts every column to the same range, around -2 to +2, where 0 means "average customer". After that, 
#all 6 columns count equally.
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled[:5]
```

```python
#create the 5 groups
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X_scaled)

df["cluster"].value_counts()
```

```python
# What this does
# groupby("cluster") puts all customers of the same group together.
# [features] looks only at our 6 behavior columns.
# mean() calculates the average of each column inside each group.
# round(1) keeps one decimal, so the numbers are easy to read.
profile = df.groupby("cluster")[features].mean().round(1)
profile
```

```python
# names is a dictionary: it says "number 0 means Low spenders, number 1 means Students", and so on.
# map(names) goes through all 5,000 customers and replaces each cluster number with its name.
# The result is saved in a new column called segment.
names = {
    0: "Low spenders",
    1: "Students",
    2: "Data lovers",
    3: "Talkers",
    4: "Premium",
}

df["segment"] = df["cluster"].map(names)
df["segment"].value_counts()
```

```python
pd.crosstab(df["true_persona"], df["segment"])
```

```python
df.to_csv("customers_segmented.csv", index=False)
```

```python
# PCA(n_components=2) creates the tool and says "give me 2 columns".
# fit_transform(X_scaled) does the squeezing for all 5,000 customers.
# The next two lines save the 2 new columns in the table as pc1 and pc2. Think of them as the x and y position of each customer on the chart.
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
points = pca.fit_transform(X_scaled)

df["pc1"] = points[:, 0]
df["pc2"] = points[:, 1]

df[["pc1", "pc2"]].head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

for name in df["segment"].unique():
    group = df[df["segment"] == name]
    plt.scatter(group["pc1"], group["pc2"], s=8, label=name)

plt.legend()
plt.title("Customer segments")
plt.xlabel("pc1")
plt.ylabel("pc2")
plt.show()
```
